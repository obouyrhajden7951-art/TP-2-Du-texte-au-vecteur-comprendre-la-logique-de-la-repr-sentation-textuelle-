# Partie 5 — Représentation Bag of Words 

```python
from sklearn.feature_extraction.text import CountVectorizer

phrases = [
    "Le football est un sport très populaire",
    "Ce film est un chef d'oeuvre du cinéma",
    "Le joueur de football regarde un film"
]
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(phrases)
```

```python
print(vectorizer.get_feature_names_out())
print(X.toarray())
```

# Partie 6 — Représentation TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

phrases = [
    "Le football est un sport très populaire",
    "Ce film est un chef d'oeuvre du cinéma",
    "Le joueur de football regarde un film"
]
```

```python
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(phrases)
```

```python
print(tfidf.get_feature_names_out())
print(X_tfidf.toarray().round(2))
```

```python
# Tableau pour comparer les poids
df = pd.DataFrame(X_tfidf.toarray().round(2),
                  columns=tfidf.get_feature_names_out(),
                  index=["Phrase 1", "Phrase 2", "Phrase 3"])
df
```
